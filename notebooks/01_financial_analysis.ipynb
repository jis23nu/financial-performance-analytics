```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("FinancialPerformanceAnalytics") \
    .getOrCreate()

spark_df = spark.read.csv(
    "../data/financial_sample_cleaned.csv",
    header=True,
    inferSchema=True
)

print("Total rows:", spark_df.count())
print("Total columns:", len(spark_df.columns))

spark_df.printSchema()
```

```python
from pyspark.sql.functions import col, sum

spark_df.select(
    [
        sum(col(c).isNull().cast("int")).alias(c)
        for c in spark_df.columns
    ]
).show()
```

```python
from pyspark.sql.functions import sum

spark_df.select(
    sum("Sales").alias("Total Sales"),
    sum("Profit").alias("Total Profit")
).show()
```

```python
from pyspark.sql.functions import sum, round, col

segment_spark = spark_df.groupBy("Segment").agg(
    sum("Sales").alias("Total Sales"),
    sum("Profit").alias("Total Profit")
)

segment_spark = segment_spark.withColumn(
    "Profit Margin (%)", 
    round(col("Total Profit")/col("Total Sales")*100,2)
)
segment_spark.show()
```

```python
segment_spark.orderBy(
    col("Total Profit").desc()
).show()
```

```python
from pyspark.sql.functions import sum, col, round

product_spark = spark_df.groupBy("Product").agg(
    sum("Sales").alias("Total Sales"),
    sum("Profit").alias("Total Profit")
)

product_spark = product_spark.withColumn(
    "Profit Margin (%)",
    round(col("Total Profit") / col("Total Sales") * 100, 2)
)

product_spark.orderBy(
    col("Total Profit").desc()
).show()
```

```python
segment_spark.toPandas().to_csv(
    "../data/segment_profitability_summary.csv",
    index=False
)

product_spark.toPandas().to_csv(
    "../data/product_profitability_summary.csv",
    index=False
)
```

```python
from pyspark.sql.functions import sum, col, round

country_spark = spark_df.groupBy("Country").agg(
    sum("Sales").alias("Total Sales"),
    sum("Profit").alias("Total Profit")
)

country_spark = country_spark.withColumn(
    "Profit Margin (%)",
    round(col("Total Profit") / col("Total Sales") * 100, 2)
)

country_spark.orderBy(
    col("Total Sales").desc()
).show()
```

```python
country_spark.toPandas().to_csv(
    "../data/country_profitability_summary.csv",
    index=False
)
```

```python
from pyspark.sql.functions import sum, col, round

monthly_spark = spark_df.groupBy("Year", "Month Number", "Month Name").agg(
    sum("Sales").alias("Total Sales"),
    sum("Profit").alias("Total Profit")
)

monthly_spark = monthly_spark.withColumn(
    "Profit Margin (%)",
    round(col("Total Profit") / col("Total Sales") * 100, 2)
)

monthly_spark.orderBy("Year", "Month Number").show(30)
```

```python
monthly_spark.toPandas().to_csv(
    "../data/monthly_financial_summary.csv",
    index=False
)
```
